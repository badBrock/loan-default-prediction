```python
import json
import numpy as np
import pandas as pd

REF_DATE = pd.Timestamp('2021-01-01')   # confirmed snapshot/application date

def load_nested_json(path):
    with open(path) as f:
        raw = json.load(f)
    records = [rec for applicant in raw for rec in applicant if rec is not None]
    return pd.DataFrame(records)
    
    # credit types worth keeping separate; rest → 'Other'
KEEP_CREDIT_TYPES = ['Consumer credit', 'Credit card', 'Car loan', 'Mortgage', 'Microloan']

def build_account_features(acc):
    acc = acc.copy()

    # --- dtypes ---
    acc['open_date']      = pd.to_datetime(acc['open_date'], errors='coerce')
    acc['closed_date']    = pd.to_datetime(acc['closed_date'], errors='coerce')
    acc['loan_amount']    = pd.to_numeric(acc['loan_amount'], errors='coerce')
    acc['amount_overdue'] = pd.to_numeric(acc['amount_overdue'], errors='coerce')

    # --- fix garbage closed_dates (closed < open → invalid) ---
    bad = acc['closed_date'] < acc['open_date']
    acc.loc[bad, 'closed_date'] = pd.NaT

    # --- derived per-row fields ---
    acc['is_open']       = acc['closed_date'].isna().astype(int)
    acc['is_zero_amt']   = (acc['loan_amount'] == 0).astype(int)
    acc['has_overdue']   = (acc['amount_overdue'] > 0).astype(int)
    acc['days_since_open']  = (REF_DATE - acc['open_date']).dt.days
    acc['days_since_close'] = (REF_DATE - acc['closed_date']).dt.days   # NaN if open
    acc['log_loan_amount']  = np.log1p(acc['loan_amount'].clip(lower=0))

    # --- grouped credit type (rare → 'Other') ---
    acc['credit_grp'] = np.where(acc['credit_type'].isin(KEEP_CREDIT_TYPES),
                                 acc['credit_type'], 'Other')

    g = acc.groupby('uid')

    # ---- NUMERICAL aggregations ----
    num = g.agg(
        acc_n_accounts       = ('uid', 'size'),
        acc_n_open           = ('is_open', 'sum'),
        acc_n_zero_amt       = ('is_zero_amt', 'sum'),

        acc_loan_amt_sum     = ('loan_amount', 'sum'),
        acc_loan_amt_mean    = ('loan_amount', 'mean'),
        acc_loan_amt_max     = ('loan_amount', 'max'),
        acc_loan_amt_std     = ('loan_amount', 'std'),
        acc_log_loan_sum     = ('log_loan_amount', 'sum'),
        acc_log_loan_mean    = ('log_loan_amount', 'mean'),

        acc_overdue_sum      = ('amount_overdue', 'sum'),
        acc_overdue_max      = ('amount_overdue', 'max'),
        acc_n_with_overdue   = ('has_overdue', 'sum'),

        acc_n_credit_types   = ('credit_type', 'nunique'),

        # recency (anchored to REF_DATE)
        acc_days_since_open_min  = ('days_since_open', 'min'),   # most recent open
        acc_days_since_open_max  = ('days_since_open', 'max'),   # oldest open
        acc_days_since_open_mean = ('days_since_open', 'mean'),
        acc_days_since_close_min = ('days_since_close', 'min'),  # most recent close
    )

    # ---- ratio features ----
    num['acc_open_ratio']       = num['acc_n_open']        / num['acc_n_accounts']
    num['acc_overdue_ratio']    = num['acc_n_with_overdue']/ num['acc_n_accounts']
    num['acc_overdue_amt_ratio']= num['acc_overdue_sum']   / (num['acc_loan_amt_sum'] + 1)
    num['acc_zero_amt_ratio']   = num['acc_n_zero_amt']    / num['acc_n_accounts']

    # ---- CATEGORICAL: count of accounts per credit-type group ----
    cat = (acc.groupby(['uid', 'credit_grp']).size()
              .unstack(fill_value=0)
              .add_prefix('acc_cnt_'))
    # normalized share per type
    cat_share = cat.div(cat.sum(axis=1), axis=0).add_suffix('_share')

    out = num.join(cat).join(cat_share)
    return out.reset_index()
    
def _parse_ph(s):
    """Return list of monthly DPD ints; most recent is last."""
    s = str(s)
    if s in ('', 'nan') or len(s) < 3:
        return []
    n = len(s) - (len(s) % 3)
    return [int(s[i:i+3]) for i in range(0, n, 3)]

def _ph_row_features(s):
    vals = _parse_ph(s)
    if not vals:
        return dict(ph_n_months=0, ph_max_dpd=0, ph_mean_dpd=0, ph_sum_dpd=0,
                    ph_n_late=0, ph_n_30=0, ph_n_90=0, ph_n_180=0, ph_n_360=0,
                    ph_recent_dpd=0, ph_recent3_max=0, ph_late_ratio=0,
                    ph_has_history=0)
    arr = np.array(vals)
    return dict(
        ph_n_months   = len(arr),
        ph_max_dpd    = int(arr.max()),
        ph_mean_dpd   = float(arr.mean()),
        ph_sum_dpd    = int(arr.sum()),
        ph_n_late     = int((arr > 0).sum()),
        ph_n_30       = int((arr >= 30).sum()),
        ph_n_90       = int((arr >= 90).sum()),
        ph_n_180      = int((arr >= 180).sum()),
        ph_n_360      = int((arr >= 360).sum()),
        ph_recent_dpd = int(arr[-1]),                 # most recent month
        ph_recent3_max= int(arr[-3:].max()),          # worst in last 3 months
        ph_late_ratio = float((arr > 0).mean()),
        ph_has_history= 1,
    )

def build_payment_features(acc):
    acc = acc.copy()
    # per-account payment features
    ph = acc['payment_hist_string'].apply(_ph_row_features).apply(pd.Series)
    ph['uid'] = acc['uid'].values

    g = ph.groupby('uid')
    out = g.agg(
        # worst-ever across all accounts
        pmt_max_dpd       = ('ph_max_dpd', 'max'),
        pmt_mean_dpd      = ('ph_mean_dpd', 'mean'),
        pmt_sum_dpd       = ('ph_sum_dpd', 'sum'),

        # delinquency counts summed across accounts
        pmt_n_late        = ('ph_n_late', 'sum'),
        pmt_n_30          = ('ph_n_30', 'sum'),
        pmt_n_90          = ('ph_n_90', 'sum'),
        pmt_n_180         = ('ph_n_180', 'sum'),
        pmt_n_360         = ('ph_n_360', 'sum'),

        # recent behaviour (max across accounts = worst recent)
        pmt_recent_dpd_max= ('ph_recent_dpd', 'max'),
        pmt_recent3_max   = ('ph_recent3_max', 'max'),

        # history coverage
        pmt_total_months  = ('ph_n_months', 'sum'),
        pmt_late_ratio_max= ('ph_late_ratio', 'max'),
        pmt_late_ratio_mean=('ph_late_ratio', 'mean'),
        pmt_n_acc_with_hist=('ph_has_history', 'sum'),
    )

    # any severe delinquency flags (very high signal, low prevalence)
    out['pmt_ever_90plus']  = (out['pmt_n_90']  > 0).astype(int)
    out['pmt_ever_180plus'] = (out['pmt_n_180'] > 0).astype(int)
    out['pmt_ever_360plus'] = (out['pmt_n_360'] > 0).astype(int)

    return out.reset_index()
    
    # only these enquiry types looked "real"; rest is likely synthetic noise
KEEP_ENQ_TYPES = ['Cash loans', 'Revolving loans']

def build_enquiry_features(enq):
    enq = enq.copy()
    enq['enquiry_date'] = pd.to_datetime(enq['enquiry_date'], errors='coerce')
    enq['enquiry_amt']  = pd.to_numeric(enq['enquiry_amt'], errors='coerce')
    enq['log_enq_amt']  = np.log1p(enq['enquiry_amt'].clip(lower=0))
    enq['days_ago']     = (REF_DATE - enq['enquiry_date']).dt.days

    # recency window flags
    for w in [30, 90, 180, 365]:
        enq[f'in_{w}d'] = (enq['days_ago'] <= w).astype(int)

    g = enq.groupby('uid')
    num = g.agg(
        enq_n              = ('uid', 'size'),
        enq_amt_sum        = ('enquiry_amt', 'sum'),
        enq_amt_mean       = ('enquiry_amt', 'mean'),
        enq_amt_max        = ('enquiry_amt', 'max'),
        enq_log_amt_sum    = ('log_enq_amt', 'sum'),
        enq_log_amt_mean   = ('log_enq_amt', 'mean'),
        enq_n_types        = ('enquiry_type', 'nunique'),

        enq_days_since_last= ('days_ago', 'min'),   # recency
        enq_days_since_first=('days_ago', 'max'),
        enq_n_30d          = ('in_30d', 'sum'),
        enq_n_90d          = ('in_90d', 'sum'),
        enq_n_180d         = ('in_180d', 'sum'),
        enq_n_365d         = ('in_365d', 'sum'),
    )
    num['enq_span_days'] = num['enq_days_since_first'] - num['enq_days_since_last']

    # CATEGORICAL: counts for the two "real" enquiry types
    enq['enq_grp'] = np.where(enq['enquiry_type'].isin(KEEP_ENQ_TYPES),
                              enq['enquiry_type'], 'Other')
    cat = (enq.groupby(['uid', 'enq_grp']).size()
              .unstack(fill_value=0)
              .add_prefix('enq_cnt_'))

    out = num.join(cat)
    return out.reset_index()
    
def build_master(flag, acc, enq):
    m = flag.copy()

    # --- flag-table feature: contract type (categorical) ---
    m['is_cash_loan'] = (m['NAME_CONTRACT_TYPE'] == 'Cash loans').astype(int)

    # --- merge aggregated features (LEFT join keeps all applicants) ---
    m = m.merge(build_account_features(acc), on='uid', how='left')
    m = m.merge(build_payment_features(acc), on='uid', how='left')
    m = m.merge(build_enquiry_features(enq), on='uid', how='left')

    # --- missingness flags BEFORE filling ---
    m['has_accounts'] = m['acc_n_accounts'].notna().astype(int)
    # (every uid has enquiries, so no has_enquiries flag needed)

    # --- fill NaNs from the LEFT join (applicants w/o accounts) ---
    acc_cols = [c for c in m.columns if c.startswith(('acc_', 'pmt_'))]
    m[acc_cols] = m[acc_cols].fillna(0)

    # enquiry cols should have no NaN (100% coverage) but be safe
    enq_cols = [c for c in m.columns if c.startswith('enq_')]
    m[enq_cols] = m[enq_cols].fillna(0)

    # std can be NaN when a group has 1 account → fill 0
    m = m.fillna(0)

    return m
    
# TRAIN
flag_train = pd.read_csv(r'E:\senior_ds_test\senior_ds_test\data\train\train_flag.csv')
acc_train  = load_nested_json(r'E:\senior_ds_test\senior_ds_test\data\train\accounts_data_train.json')
enq_train  = load_nested_json(r'E:\senior_ds_test\senior_ds_test\data\train\enquiry_data_train.json')
train = build_master(flag_train, acc_train, enq_train)

# TEST
flag_test  = pd.read_csv(r'E:\senior_ds_test\senior_ds_test\data\test\test_flag.csv')   # or sample_submission
acc_test   = load_nested_json(r'E:\senior_ds_test\senior_ds_test\data\test\accounts_data_test.json')
enq_test   = load_nested_json(r'E:\senior_ds_test\senior_ds_test\data\test\enquiry_data_test.json')
test  = build_master(flag_test, acc_test, enq_test)

print("Train:", train.shape)
print("Test :", test.shape)

# ---- align columns between train & test ----
# (one-hot / count columns can differ if a category appears in one set only)
feature_cols = [c for c in train.columns
                if c not in ['uid', 'TARGET', 'NAME_CONTRACT_TYPE']]

# add missing cols to test (fill 0), drop extras
for c in feature_cols:
    if c not in test.columns:
        test[c] = 0

# ensure test only carries the same feature set, in the same order
test = test[['uid'] + feature_cols]

print("\nFinal feature count:", len(feature_cols))
print("Any NaN in train features?", train[feature_cols].isna().any().any())
print("Any NaN in test features? ", test[feature_cols].isna().any().any())


# ---- 6a. shape & no leakage columns ----
assert 'TARGET' not in feature_cols
assert train['uid'].is_unique
assert test['uid'].is_unique

# ---- 6b. do our "strong" payment features separate the classes? ----
check_cols = [
    'pmt_max_dpd', 'pmt_n_90', 'pmt_n_360', 'pmt_ever_90plus',
    'acc_n_with_overdue', 'acc_open_ratio', 'acc_n_accounts',
    'enq_n', 'enq_n_90d', 'enq_days_since_last', 'is_cash_loan',
    'has_accounts'
]
print(train.groupby('TARGET')[check_cols].mean().T)

corr = train[feature_cols + ['TARGET']].corr()['TARGET'].drop('TARGET')
print("TOP + correlated with default:\n", corr.sort_values(ascending=False).head(15))
print("\nTOP - correlated with default:\n", corr.sort_values().head(10))

train.to_csv('features_train.csv', index=False)
test.to_csv('features_test.csv', index=False)
print("Saved engineered features as CSV.")
print("Saved engineered features.")
```

```python
def add_interaction_features(df):
    df = df.copy()
    eps = 1e-6
    # credit-hunger: recent enquiries relative to total history
    df['enq_recent_ratio']    = df['enq_n_90d'] / (df['enq_n'] + eps)
    df['enq_velocity']        = df['enq_n'] / (df['enq_span_days'] + 30)
    # burden: open loans vs total, amount per account
    df['open_x_amount']       = df['acc_open_ratio'] * df['acc_log_loan_mean']
    df['amt_per_account']     = df['acc_log_loan_sum'] / (df['acc_n_accounts'] + eps)
    # tenure gap: how long since most recent vs oldest account
    df['acc_tenure_gap']      = df['acc_days_since_open_max'] - df['acc_days_since_open_min']
    # enquiry vs account activity mismatch (applying a lot but few accounts = risk)
    df['enq_to_acc_ratio']    = df['enq_n'] / (df['acc_n_accounts'] + 1)
    # recency combo
    df['recency_combo']       = df['acc_days_since_open_mean'] + df['enq_days_since_first']
    # microloan risk amplified by open ratio
    df['microloan_x_open']    = df['acc_cnt_Microloan_share'] * df['acc_open_ratio']
    return df

train = add_interaction_features(train)
test  = add_interaction_features(test)

feature_cols = [c for c in train.columns
                if c not in ['uid','TARGET','NAME_CONTRACT_TYPE']]
# align test again
for c in feature_cols:
    if c not in test.columns: test[c] = 0
test = test[['uid'] + feature_cols]

print("New feature count:", len(feature_cols))

X = train[feature_cols]; y = train['TARGET']
oof = cross_val_predict(model, X, y, cv=cv, method='predict_proba', n_jobs=-1)[:,1]
print("CV AUC with interactions:", round(roc_auc_score(y, oof), 5))
```

```python
import optuna
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import roc_auc_score
from xgboost import XGBClassifier

X = train[feature_cols]
y = train['TARGET']
spw = (y == 0).sum() / (y == 1).sum()
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

def objective(trial):
    params = dict(
        n_estimators     = trial.suggest_int('n_estimators', 300, 900),
        max_depth        = trial.suggest_int('max_depth', 3, 6),
        learning_rate    = trial.suggest_float('learning_rate', 0.01, 0.1, log=True),
        subsample        = trial.suggest_float('subsample', 0.6, 1.0),
        colsample_bytree = trial.suggest_float('colsample_bytree', 0.5, 1.0),
        min_child_weight = trial.suggest_int('min_child_weight', 1, 20),   # ↑ = more reg
        gamma            = trial.suggest_float('gamma', 0, 5),             # min split gain
        reg_alpha        = trial.suggest_float('reg_alpha', 0, 5),         # L1
        reg_lambda       = trial.suggest_float('reg_lambda', 0, 10),       # L2
        scale_pos_weight = spw,
        eval_metric      = 'auc',
        n_jobs           = -1,
        random_state     = 42,
    )
    model = XGBClassifier(**params)
    oof = cross_val_predict(model, X, y, cv=cv, method='predict_proba', n_jobs=-1)[:, 1]
    return roc_auc_score(y, oof)

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=40, show_progress_bar=True)

print("\n📊 BEST CV AUC:", round(study.best_value, 5))
print("BEST PARAMS:", study.best_params)
```

```python
# ============================================================
#  MULTI-MODEL COMPARISON + ENSEMBLE
# ============================================================
import numpy as np
from scipy.stats import rankdata
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import roc_auc_score
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

# ---- data ----
X = train[feature_cols]
y = train['TARGET']
spw = float((y == 0).sum() / (y == 1).sum())   # plain float (avoids CatBoost clone bug)
print("scale_pos_weight:", round(spw, 2))

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# ============================================================
# 1. XGBOOST (best tuned params from Optuna)
# ============================================================
xgb_params = {
    'n_estimators': 481, 'max_depth': 3, 'learning_rate': 0.05961,
    'subsample': 0.74043, 'colsample_bytree': 0.64632,
    'min_child_weight': 8, 'gamma': 3.85739,
    'reg_alpha': 3.66946, 'reg_lambda': 8.80245,
    'scale_pos_weight': spw, 'eval_metric': 'auc',
    'random_state': 42, 'n_jobs': -1,
}
xgb = XGBClassifier(**xgb_params)
oof_xgb = cross_val_predict(xgb, X, y, cv=cv, method='predict_proba', n_jobs=-1)[:, 1]
print("XGBoost  CV AUC:", round(roc_auc_score(y, oof_xgb), 5))

# ============================================================
# 2. LIGHTGBM
# ============================================================
lgb = LGBMClassifier(
    n_estimators=500, max_depth=3, num_leaves=8,
    learning_rate=0.05, subsample=0.75, colsample_bytree=0.65,
    reg_alpha=3.5, reg_lambda=8.5, min_child_samples=40,
    scale_pos_weight=spw, random_state=42, n_jobs=-1, verbose=-1
)
oof_lgb = cross_val_predict(lgb, X, y, cv=cv, method='predict_proba', n_jobs=-1)[:, 1]
print("LightGBM CV AUC:", round(roc_auc_score(y, oof_lgb), 5))

# ============================================================
# 3. CATBOOST  (n_jobs=1 -> CatBoost handles its own threads)
# ============================================================
cat = CatBoostClassifier(
    iterations=500, depth=3, learning_rate=0.05,
    l2_leaf_reg=8, subsample=0.75,
    scale_pos_weight=spw, random_state=42,
    verbose=0, allow_writing_files=False
)
oof_cat = cross_val_predict(cat, X, y, cv=cv, method='predict_proba', n_jobs=1)[:, 1]
print("CatBoost CV AUC:", round(roc_auc_score(y, oof_cat), 5))

# ============================================================
# 4. ENSEMBLE
# ============================================================
print("\n---- model correlations (lower = more diverse) ----")
print("corr xgb-lgb:", round(np.corrcoef(oof_xgb, oof_lgb)[0, 1], 4))
print("corr xgb-cat:", round(np.corrcoef(oof_xgb, oof_cat)[0, 1], 4))
print("corr lgb-cat:", round(np.corrcoef(oof_lgb, oof_cat)[0, 1], 4))

# simple average
blend = (oof_xgb + oof_lgb + oof_cat) / 3
print("\n🎯 BLEND      CV AUC:", round(roc_auc_score(y, blend), 5))

# rank average (usually best for AUC since AUC is rank-based)
rank_blend = (rankdata(oof_xgb) + rankdata(oof_lgb) + rankdata(oof_cat)) / 3
print("🎯 RANK-BLEND CV AUC:", round(roc_auc_score(y, rank_blend), 5))

# ---- summary table ----
print("\n================ SUMMARY ================")
for name, oof in [('XGBoost', oof_xgb), ('LightGBM', oof_lgb),
                  ('CatBoost', oof_cat), ('Blend', blend),
                  ('Rank-Blend', rank_blend)]:
    print(f"{name:12s}: {roc_auc_score(y, oof):.5f}")
```

```python
# ============================================================
#  OPTIMIZATION: feature selection + spw sweep + ensemble + submission
# ============================================================
import numpy as np, pandas as pd, os
from scipy.stats import rankdata
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import roc_auc_score
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

# make sure we're on the clean 69 features
train = pd.read_csv('features_train.csv')
test  = pd.read_csv('features_test.csv')
feature_cols = [c for c in train.columns if c not in ['uid','TARGET','NAME_CONTRACT_TYPE']]
for c in feature_cols:
    if c not in test.columns: test[c] = 0
test = test[['uid'] + feature_cols]

y   = train['TARGET']
spw = float((y==0).sum()/(y==1).sum())
cv  = StratifiedKFold(5, shuffle=True, random_state=42)

BEST = dict(n_estimators=481, max_depth=3, learning_rate=0.05961,
            subsample=0.74043, colsample_bytree=0.64632,
            min_child_weight=8, gamma=3.85739,
            reg_alpha=3.66946, reg_lambda=8.80245,
            eval_metric='auc', random_state=42, n_jobs=-1)

def cv_auc(params, cols):
    m = XGBClassifier(**params)
    oof = cross_val_predict(m, train[cols], y, cv=cv,
                            method='predict_proba', n_jobs=-1)[:,1]
    return roc_auc_score(y, oof), oof

# ---------- STEP 1: scale_pos_weight sweep ----------
print("STEP 1 — spw sweep")
best_spw, best_auc = spw, 0
for w in [1.0, spw**0.5, spw*0.5, spw]:
    a,_ = cv_auc({**BEST,'scale_pos_weight':w}, feature_cols)
    print(f"  spw={w:6.2f}  AUC={a:.5f}")
    if a > best_auc: best_auc, best_spw = a, w
BEST['scale_pos_weight'] = best_spw
print(f"  -> best spw={best_spw:.2f}  AUC={best_auc:.5f}\n")

# ---------- STEP 2: feature selection ----------
print("STEP 2 — feature selection")
m = XGBClassifier(**BEST).fit(train[feature_cols], y)
imp = pd.Series(m.feature_importances_, index=feature_cols)
base_auc,_ = cv_auc(BEST, feature_cols)
print(f"  all {len(feature_cols)} feats: AUC={base_auc:.5f}")
best_cols, best_sel = feature_cols, base_auc
for q in [0.10, 0.20, 0.30]:
    keep = imp[imp > imp.quantile(q)].index.tolist()
    a,_ = cv_auc(BEST, keep)
    print(f"  drop bottom {int(q*100)}% -> {len(keep)} feats: AUC={a:.5f}")
    if a > best_sel: best_sel, best_cols = a, keep
print(f"  -> kept {len(best_cols)} feats  AUC={best_sel:.5f}\n")

# ---------- STEP 3: 3-model rank-blend on chosen config ----------
print("STEP 3 — ensemble on selected features")
Xtr = train[best_cols]

xgb = XGBClassifier(**BEST)
oof_x = cross_val_predict(xgb, Xtr, y, cv=cv, method='predict_proba', n_jobs=-1)[:,1]

lgb = LGBMClassifier(n_estimators=500, max_depth=3, num_leaves=8,
        learning_rate=0.05, subsample=0.75, colsample_bytree=0.65,
        reg_alpha=3.5, reg_lambda=8.5, min_child_samples=40,
        scale_pos_weight=best_spw, random_state=42, n_jobs=-1, verbose=-1)
oof_l = cross_val_predict(lgb, Xtr, y, cv=cv, method='predict_proba', n_jobs=-1)[:,1]

cat = CatBoostClassifier(iterations=500, depth=3, learning_rate=0.05,
        l2_leaf_reg=8, subsample=0.75, scale_pos_weight=best_spw,
        random_state=42, verbose=0, allow_writing_files=False)
oof_c = cross_val_predict(cat, Xtr, y, cv=cv, method='predict_proba', n_jobs=1)[:,1]

print(f"  XGB={roc_auc_score(y,oof_x):.5f}  LGB={roc_auc_score(y,oof_l):.5f}  CAT={roc_auc_score(y,oof_c):.5f}")
rank_blend = (rankdata(oof_x)+rankdata(oof_l)+rankdata(oof_c))/3
blend_auc  = roc_auc_score(y, rank_blend)
print(f"  RANK-BLEND AUC={blend_auc:.5f}")

# pick the best of {xgb-alone, blend}
use_blend = blend_auc > roc_auc_score(y, oof_x)
final_auc = blend_auc if use_blend else roc_auc_score(y, oof_x)
print(f"\n🏆 FINAL CV AUC = {final_auc:.5f}  ({'rank-blend' if use_blend else 'xgb-only'})")

# ---------- STEP 4: fit on full data, predict test, submission ----------
print("\nSTEP 4 — submission")
xgb_f = XGBClassifier(**BEST).fit(Xtr, y)
pred_x = xgb_f.predict_proba(test[best_cols])[:,1]

if use_blend:
    lgb_f = lgb.fit(Xtr, y); pred_l = lgb_f.predict_proba(test[best_cols])[:,1]
    cat_f = cat.fit(Xtr, y); pred_c = cat_f.predict_proba(test[best_cols])[:,1]
    test_pred = (rankdata(pred_x)+rankdata(pred_l)+rankdata(pred_c))/(3*len(pred_x))
else:
    test_pred = pred_x

sub = pd.DataFrame({'uid': test['uid'], 'TARGET': test_pred})
out_dir = r'E:\senior_ds_test\senior_ds_test\final_submission'
os.makedirs(out_dir, exist_ok=True)
out_path = os.path.join(out_dir, 'final_submission_Sakshi_Vedi.csv')
sub.to_csv(out_path, index=False)
print("  saved:", out_path)
print(sub.head())
```
